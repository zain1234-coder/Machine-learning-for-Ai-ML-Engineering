```python
import pandas as pd
import numpy as np
import seaborn  as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer
```

```python
df = pd.read_csv("train.csv" , usecols=['Age' , 'Survived' , 'Fare'])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean()) 
```

```python
df
```

```python
x = df.iloc[:,1:]
y = df.iloc[:,0]
```

```python
x_train , x_test , y_train , y_test =train_test_split(x , y ,test_size=0.2 , random_state=0)
```

# Checking Age and Fare olumns is normally distributed or not

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
sns.histplot(x_train['Age'], kde=True)
plt.title("Age Pdf")

plt.subplot(122)
stats.probplot(x_train['Age'] , dist='norm',plot=plt)
plt.title("QQ Plot")

plt.show()
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
sns.histplot(x_train['Fare'],kde=True)
plt.title("Fare pdf")

plt.subplot(122)
stats.probplot(x_train['Fare'], dist='norm' ,plot=plt)
plt.title("QQ Plot")

plt.show()
```

```python
clf = LogisticRegression()
clf2 = DecisionTreeClassifier()
```

```python
clf.fit(x_train , y_train)
```

```python

clf2.fit(x_train,y_train)
```

```python
y_pred = clf.predict(x_test)
y_pred2 = clf2.predict(x_test)
```

```python
print("Accuracy LG" , accuracy_score(y_test , y_pred))
print("Accuracy DT" , accuracy_score(y_test,y_pred2))
```

# With function  Transformation

```python
trf1 = FunctionTransformer(func = np.log1p)
```

```python
x_train_transformed = trf1.transform(x_train)
x_test_transformed = trf1.transform(x_test)
```

```python
clf.fit(x_train_transformed,y_train)
clf2.fit(x_train_transformed,y_train)
```

```python
y_pred3 = clf.predict(x_test)
y_pred4 = clf2.predict(x_test)
```

```python
print("Accuracy lg" , accuracy_score(y_test,y_pred3))
print("Accuracy Dt" , accuracy_score(y_test , y_pred4))
```

```python
x_tranformed = trf1.fit_transform(x)

clf = LogisticRegression()
clf2 = DecisionTreeClassifier()

print("LR" , np.mean(cross_val_score(clf,x_tranformed,y,scoring='accuracy',cv=10)))
print("DT" , np.mean(cross_val_score(clf2,x_tranformed,y,scoring='accuracy',cv=10)))
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
sns.histplot(x_train['Fare'] ,kde=True)
plt.title("Before Transform Fare")

plt.subplot(122)
sns.histplot(x_train_transformed['Fare'] , kde=True)
plt.title("After Transform Fare")

plt.show()
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
stats.probplot(x_train['Age'] , dist='norm' , plot=plt)
plt.title("Before Transform Age")

plt.subplot(122)
stats.probplot(x_train_transformed['Age'] , dist='norm' , plot=plt)
plt.title("After Transform Age")

plt.show()
```

```python
trf2 = ColumnTransformer([('log',FunctionTransformer(np.log1p),['Fare'])],remainder='passthrough')

X_train_transformed2 = trf2.fit_transform(x_train)
X_test_transformed2 = trf2.transform(x_test)
```

```python
clf = LogisticRegression()
clf2 = DecisionTreeClassifier()

clf.fit(X_train_transformed2,y_train)
clf2.fit(X_train_transformed2,y_train)
    
y_pred = clf.predict(X_test_transformed2)
y_pred2 = clf2.predict(X_test_transformed2)
    
print("Accuracy LR",accuracy_score(y_test,y_pred))
print("Accuracy DT",accuracy_score(y_test,y_pred2))
```

```python
X_transformed2 = trf2.fit_transform(x)

clf = LogisticRegression()
clf2 = DecisionTreeClassifier()

print("LR",np.mean(cross_val_score(clf,X_transformed2,y,scoring='accuracy',cv=10)))
print("DT",np.mean(cross_val_score(clf2,X_transformed2,y,scoring='accuracy',cv=10)))
```

```python
def apply_transform(transform):
    X = df.iloc[:,1:3]
    y = df.iloc[:,0]
    
    trf = ColumnTransformer([('log',FunctionTransformer(transform),['Fare','Age'])],remainder='passthrough')
    
    X_trans = trf.fit_transform(X)
    
    clf = LogisticRegression()
    
    print("Accuracy",np.mean(cross_val_score(clf,X_trans,y,scoring='accuracy',cv=10)))
    
    plt.figure(figsize=(14,4))

    plt.subplot(121)
    stats.probplot(X['Fare'], dist="norm", plot=plt)
    plt.title('Fare Before Transform')

    plt.subplot(122)
    stats.probplot(X_trans[:,0], dist="norm", plot=plt)
    plt.title('Fare After Transform')

    plt.show()
```

```python
apply_transform(lambda X : X**2)
```

```python
apply_transform(lambda X : X**1/2)
```

```python
apply_transform(lambda X : 1/(X+0.0001)) # x+0.1 so our dnominator never goes zero
```
