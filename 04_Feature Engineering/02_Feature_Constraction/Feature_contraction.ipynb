```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
cross_val_score(LogisticRegression(),x,y, scoring='accuracy' , cv =20).mean()
```

# Applying Feature Constraction

```python
x['Family_type'] = x['SibSp'] + x['Parch'] +1
```

```python
x
```

```python
def myfunc (num):
    if num == 1 :
        return 0 
    if num > 1 & num <=4 :
        return 1
    else:
        return 2
```

```python
x['Family_type'] = x['Family_type'].apply(myfunc)
```

```python
x = x.drop(columns= ['SibSp' , 'Parch'])
```

```python
x
```

```python
cross_val_score(LogisticRegression() , x,y , scoring='accuracy' , cv = 20) .mean()
```
