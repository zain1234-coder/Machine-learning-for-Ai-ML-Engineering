```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9)
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df_cop = df[['R&D Spend','Administration','Marketing Spend']].copy()
```

```python
df.iloc[3,1] # 3 for row and 1 for value
```

```python
df_cop.iloc[1,0] = np.nan
df_cop.iloc[3,1] = np.nan
df_cop.iloc[-1,-1] = np.nan
df_cop.head()
```

# **Step1 impute all values with mean of that col**

```python
df0 = pd.DataFrame()

df0['R&D Spend'] =df_cop['R&D Spend'].fillna(df_cop['R&D Spend'].mean())
df0['Administration'] = df_cop['Administration'].fillna(df_cop['Administration'].mean())
df0['Marketing Spend'] = df_cop['Marketing Spend'].fillna(df_cop['Marketing Spend'].mean())
```

# **Remove the col2 imputed value**

```python
df1 = df0.copy()
df1.iloc[1,0] = np.nan
df1
```

```python
# Use first 3 rows to build a model and use the last for prediction

x = df1.iloc[[0,2,3,4],1:]
y = df1.iloc[[0,2,3,4],0]
```

```python
lr = LinearRegression()

lr.fit(x,y)

lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 23.14
```
