```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv' ,usecols=['Age','Pclass','Fare','Survived'])
```

```python
df.head()

df.isnull().mean() * 100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train.head()
```

```python
knn = KNNImputer(n_neighbors=5 , weights= 'uniform') # k value = 5 , k is n_neighbors , we can pass uniform and distance in weights

x_train_knn = knn.fit_transform(X_train)
x_test_knn = knn.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(x_train_knn,y_train)

y_pred = lr.predict(x_test_knn)

accuracy_score(y_test,y_pred)
```
