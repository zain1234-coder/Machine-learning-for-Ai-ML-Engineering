```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('houses.csv',usecols=['GarageQual' , 'FireplaceQu' , 'SalePrice'])
df.isnull().mean()*100
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
df['GarageQual'].value_counts(ascending=False).plot(kind='bar')
plt.title("GarageQual")

plt.subplot(122)
df['FireplaceQu'].value_counts().plot(kind='bar')
plt.title("FireplaceQu")
plt.show()
```

```python
df['GarageQual'] = df['GarageQual'].fillna('Missing')
df['FireplaceQu'] = df['FireplaceQu'].fillna('Misssing')
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
df['GarageQual'].value_counts(ascending=False).plot(kind='bar')
plt.title("GarageQual")

plt.subplot(122)
df['FireplaceQu'].value_counts().plot(kind='bar')
plt.title("FireplaceQu")
plt.show()
```

# With Sklearn

```python
from sklearn.model_selection import train_test_split
x_train , x_test ,y_train ,y_test =train_test_split(df.drop(columns=['SalePrice']),
                                                  df['SalePrice'],
                                                  test_size=0.2,
                                                  random_state=2)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='constant' , fill_value='Missing')
```

```python
x_train = imputer.fit_transform(x_train)
x_test = imputer.transform(x_test)
```

```python
x_train = pd.DataFrame(x_train , columns = imputer.get_feature_names_out())
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
x_train['GarageQual'].value_counts(ascending=False).plot(kind='bar')
plt.title("GarageQual")

plt.subplot(122)
x_train['FireplaceQu'].value_counts().plot(kind='bar')
plt.title("FireplaceQu")
plt.show()
```
