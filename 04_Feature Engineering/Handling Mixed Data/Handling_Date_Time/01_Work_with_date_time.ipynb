```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date['date'].info()
```

```python
# Extracting year

date['year'] = date['date'].dt.year #dt = datetime
```

```python
# extracting mont

date['Month'] = date['date'].dt.month
```

```python
# extracting month name

date['month_name'] = date['date'].dt.month_name()
```

```python
# extracting days 

date['days'] = date['date'].dt.day
```

```python
date['day_name'] = date['date'].dt.day_name()
```

```python
# day of week
date['day_of_week'] = date['date'].dt.day_of_week
```

```python
# day_of_year

date['day_of_year'] = date['date'].dt.day_of_week
```

```python
date['Weekend'] = np.where(date['day_name'].isin(['Sunday' , 'Saturday']),1,0)
```

```python
date['Week_of_year'] = date['date'].dt.isocalendar().week
```

```python
date['Quater'] = date['date'].dt.quarter
```

```python
date['Semester'] = np.where(date['Quater'].isin([1,2]) , 1 ,2)
```

```python
todays - date['date'] 
```

```python
# passed days

(todays - date['date']).dt.days
```
