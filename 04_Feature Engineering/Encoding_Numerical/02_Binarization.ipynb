```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.tree import DecisionTreeClassifier

from sklearn.compose import ColumnTransformer
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv',usecols=['Age' , 'Fare','SibSp' ,'Parch',"Survived"])
```

```python
df['family'] = df['SibSp']+df['Parch']
```

```python
df.head()
```

```python
df = df.drop(columns=['SibSp','Parch'])
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2 , random_state=42)

x_train
```

```python
# Without binarization

clf = DecisionTreeClassifier()

clf.fit(x_train,y_train)

y_pred = clf.predict(x_test)

print(accuracy_score(y_test,y_pred))
```

```python
cross_val_score(clf,x,y,scoring='accuracy',cv=10).mean()
```

# Binarization

```python
from sklearn.preprocessing import Binarizer
```

```python
trf = ColumnTransformer([
    ('binarize' , Binarizer(copy=False),['family'])
],remainder='passthrough')
```

```python
x_train_trf = trf.fit_transform(x_train)
x_test_trf = trf.transform(x_test)
```

```python
pd.DataFrame(x_train_trf,columns=['family' , 'Age' , 'Fare'])
```

```python
# Without binarization

clf = DecisionTreeClassifier()

clf.fit(x_train_trf,y_train)

y_pred3 = clf.predict(x_test_trf)

print(accuracy_score(y_test,y_pred3))
```

```python
x_trfx = trf.fit_transform(x)
cross_val_score(clf,x_trfx,y,scoring='accuracy' , cv=10).mean()
```
