```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")
```

```python
df.head()
```

# 1. Categorical data 

- for frequency count we use count plot

#### HOW MANY SRVIVED

```python
sns.countplot(x = df['Survived']) # basic count 
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x = df['Pclass'])
```

### For percentage count use pie plot

```python
df['Sex'].value_counts().plot(kind='pie',autopct= '%.2f')
```

# 2. Numerical data

### a) Histogram   ---->> For checking distribution of data

```python
plt.hist(df['Age'] , bins = [10,20,40,50,60,70,80,90])
plt.show()
```

## b) Distplot

```python
sns.displot(data=df, x='Age', kde=True)
plt.show()
```

### c) boxplot

```python
sns.boxplot(x = df["Fare"])
```

```python
sns.boxplot(x = df["Age"])
```
