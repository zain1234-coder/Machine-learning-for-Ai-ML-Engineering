# Skiprows/nrows Parameter

# Convertors

```python
def rename (name):
    if name == "Royal Challengers Bangalore" :
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python
pd.read_csv("ipl-matches.csv",converters={"Team1":rename}) # Change the name in team 1 Royal Challengers Bangalore to RCB
```
