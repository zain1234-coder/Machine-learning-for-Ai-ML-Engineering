# Rapid API For exploring making new datasets remember it to upload on kaggel

```python
import pandas as pd
import requests

api_key = 'aa928beda6416b52d792f413ebf7a7b1'
all_pages = []

for i in range(1, 429):
    try:
        response = requests.get(
            'https://api.themoviedb.org/3/movie/top_rated'
            '?api_key={}&language=en-US&page={}'.format(api_key, i),
            timeout=10
        )
        data = response.json()

        if 'results' not in data:
            print(f"Page {i} failed: {data}")
            continue

        temp_df = pd.DataFrame(data['results'])[
            ['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']
        ]
        all_pages.append(temp_df)

        if i % 20 == 0:
            print(f"Page {i} done")  # progress check

    except requests.exceptions.RequestException as e:
        print(f"Page {i} error: {e}")
        continue

df = pd.concat(all_pages, ignore_index=True)
```
