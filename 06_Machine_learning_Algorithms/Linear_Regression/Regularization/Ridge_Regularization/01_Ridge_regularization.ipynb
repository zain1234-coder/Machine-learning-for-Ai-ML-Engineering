```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_diabetes
import matplotlib.pyplot as plt
```

```python
data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
x = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test , y_train,y_test = train_test_split(x,y,test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
Lr = LinearRegression()

Lr.fit(x_train,y_train)
```

```python
y_predict = Lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score , mean_squared_error

print('r2 score',r2_score(y_test,y_predict))
print('Root mean square eerror',np.sqrt(mean_squared_error(y_test,y_predict)))
```

# **Ridge Regression**

```python
from sklearn.linear_model import Ridge
```

```python
Rg = Ridge(alpha=0.0001) # alpha is basically the value of lamba
```

```python
Rg.fit(x_train,y_train)
```

```python
Rg_pred = Rg.predict(x_test)
```

```python
print('r2 score',r2_score(y_test,Rg_pred))
print('Root mean square eerror',np.sqrt(mean_squared_error(y_test,Rg_pred)))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import Ridge  # <-- Ye line add karna zaroori thi
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

# 1. Ek function banaya jo Ridge Regression model train karega
def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        # Degree=16 ka matlab hai ye model data ko seekhne ke liye bohot zyaada tedhi (complex) lines bana sakta hai
        ('poly_feats', PolynomialFeatures(degree=16)),
        # Ridge model overfitting ko rokta hai. Isme 'alpha' batata hai ke model par kitni sakhti karni hai
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)      # Model ko x1 aur x2 data de kar train kiya
    return model.predict(x1)  # Train hone ke baad model se predictions (answers) liye

# 2. Teeno alag-alag alpha values aur unke colors set kiye
alphas = [0, 20, 200]  # Alpha=0 matlab koi control nahi (overfitting), Alpha=200 matlab bohot zyada sakhti (underfitting)
cs = ['r', 'g', 'b']   # Har line ka rang: Red, Green, aur Blue

plt.figure(figsize=(10, 6))  # Graph ka size set kiya
plt.plot(x1, x2, 'b+', label='Datapoints')  # Asal data points ko Blue color ke '+' signs se show kiya

# 3. Loop chalaya taake teeno alphas par check kar sakein
for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)  # Function ko call karke predictions nikali
    
    # Graph par line hamesha hilti hui na dikhe, isliye X-axis ko 'sorted' (tarteeb) me plot karna zaroori hota hai
    # np.argsort(x1[:, 0]) se predictions bhi usi tarteeb me aa jati hain jis tarteeb me X sorted hai
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

# 4. Graph ko screen par show karna
plt.legend()  # Konse rang ki line kis alpha ki hai, wo box show karega
plt.show()    # Graph screen par display karega
```
